```python
import requests
import json
import re

def get_species_languages(pokemon_name):
    url = f"https://pokeapi.co/api/v2/pokemon-species/{pokemon_name.lower()}/"
    data = requests.get(url).json()

    # Extract all language → name pairs
    translations = {
        entry["language"]["name"]: entry["name"]
        for entry in data["names"]
    }
    return translations

def get_moves_languages(pokemon_name):
    url = f"https://pokeapi.co/api/v2/move/{pokemon_name.lower()}/"
    data = requests.get(url).json()

    # Extract all language → name pairs
    translations = {
        entry["language"]["name"]: entry["name"]
        for entry in data["names"]
    }
    return translations
```

```python
spelling_mistakes = {
    "futuresight": "future-sight",
    "super_power": "superpower",
    "pyroball": "pyro-ball",
    "myst_fire": "mystical-fire",
}
MOVE_REGEX_MAP = {
    r"^weather_ball_[a-z]+$": "weather-ball",
    r"^techno_blast_[a-z]+$": "techno-blast",
    r"^aura_wheel_[a-z]+$": "aura-wheel",
}

def fix_move_name(move_name):
    if move_name in spelling_mistakes:
        move_name = spelling_mistakes[move_name]
    for pattern, base in MOVE_REGEX_MAP.items():
        if re.match(pattern, move_name):
            return base
    return move_name  # fallback: already a normal move
```

```python
with open("game_master2.json", "r") as f:
    file_contents = json.load(f)

moves = [entry for entry in file_contents if 'combatMove' in entry['data']]
error_names = []
for i,move in enumerate(moves):
    name = move["data"]["combatMove"]["name"]
    name = fix_move_name(name)
    print(f"Translating: {name}; Move {i}/{len(moves)}")
    try:
        print(get_moves_languages(name.replace('_fast','').replace('_','-')))
    except:
        error_names.append(name)
        pass
print(error_names)
```
